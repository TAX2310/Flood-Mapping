```python
from pathlib import Path

def get_leaf_subdirs(root_dir):
    root_dir = Path(root_dir)

    return [
        p for p in root_dir.rglob("*")
        if p.is_dir() and not any(child.is_dir() for child in p.iterdir())
    ]

subdirs = get_leaf_subdirs(cfg.EXP_DIR)
print(subdirs)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

subdirs = get_leaf_subdirs("/content/drive/MyDrive/MSc/Flood-Mapping/experiments")
print(subdirs)

dropdown = widgets.Dropdown(
    options=subdirs,
    value=subdirs[0],
    description="DATA_TYPE:"
)

run_button = widgets.Button(description="Plot metrics")
output = widgets.Output()

def on_click(button):
    with output:
        clear_output(wait=True)
        plotting.plot_metrics(cfg.EXP_DIR / dropdown.value)

run_button.on_click(on_click)

display(dropdown, run_button, output)
```
